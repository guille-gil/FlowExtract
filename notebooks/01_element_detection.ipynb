# Element Detection Exploration

This notebook explores the element detection stage of the pipeline.

```python
import sys
sys.path.append('..')

import cv2
import matplotlib.pyplot as plt
from pathlib import Path

from src.utils.io_utils import load_config
from src.detection.detector import ElementDetector

%matplotlib inline
```

## Load Configuration

```python
config = load_config('../configs/pipeline_config.yaml')
print("Detection config:", config['detection'])
```

## Test on Single Image

```python
# Initialize detector
detector = ElementDetector(config)

# Select test image
test_image_path = '../data/input/tocaps/page_001.png'  # Adjust path as needed

# Process image
result = detector.process_image(
    test_image_path,
    output_dir='../outputs/test_detection',
    save_visualization=True
)

print(f"Detected {result['num_elements']} elements")
```

## Visualize Results

```python
# Load original and visualization
original = cv2.imread(test_image_path)
original = cv2.cvtColor(original, cv2.COLOR_BGR2RGB)

vis_path = '../outputs/test_detection/' + Path(test_image_path).stem + '_detection_vis.png'
visualization = cv2.imread(vis_path)
visualization = cv2.cvtColor(visualization, cv2.COLOR_BGR2RGB)

# Display
fig, axes = plt.subplots(1, 2, figsize=(15, 8))
axes[0].imshow(original)
axes[0].set_title('Original')
axes[0].axis('off')

axes[1].imshow(visualization)
axes[1].set_title(f'Detected Elements ({result["num_elements"]})')
axes[1].axis('off')

plt.tight_layout()
plt.show()
```

Experiment with different parameters to improve detection quality.
